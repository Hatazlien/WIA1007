```python
df = pd.read_csv("/content/drive/MyDrive/studentrecord.csv")

# display the first 5 rows
print("First 5 rows:")
print(df.head())

# research score vs project score
sns.scatterplot(
    data=df,
    x="ResearchScore",
    y="ProjectScore",
    hue="Recommend"
)
plt.title("Research Score vs Project Score")
plt.show()

# recommendation counts
sns.countplot(data=df, x="Recommend")
plt.title("Number of Recommended Students")
plt.show()
```
