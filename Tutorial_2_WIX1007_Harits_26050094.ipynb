```python
import numpy as np

array_2x2 = np.array([[1, 0], [0, 1]])

print("1. \nNumPy array of size 2 x 2:")
print(array_2x2)
```

```python
array_random = np.random.rand(3, 4)

print("2. \nRandom NumPy array of size 3 x 4:")
print(array_random)
```

```python
array_ones = np.ones((3, 3))
array_zeros = np.zeros((3, 3))

print("3. \nArray of ones:")
print(array_ones)

print("\nArray of zeros:")
print(array_zeros)
```

```python
array_6x3 = np.random.rand(6, 3)

print("4. \nOriginal array (6 x 3):")
print(array_6x3)

array_transposed = array_6x3.T

print("\nTransposed array (3 x 6):")
print(array_transposed)
```

```python
n = 5
p = 3
m = 6

array_A = np.random.rand(n, p)
array_B = np.random.rand(p, m)

print("5. \nMatrix A (n x p):")
print(array_A)

print("\nMatrix B (p x m):")
print(array_B)

dot_product = np.dot(array_A, array_B)

print("\nDot product (n x m):")
print(dot_product)
```
